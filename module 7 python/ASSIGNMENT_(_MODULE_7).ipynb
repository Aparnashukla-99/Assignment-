```python
# 20) Swap first two characters of two strings
def swap_first_two(str1, str2):
    new_str1 = str2[:2] + str1[2:]
    new_str2 = str1[:2] + str2[2:]
    return new_str1 + " " + new_str2

print(swap_first_two("abc", "xyz"))
```

```python
# 21) Add "ing"/"ly" or leave unchanged
def add_string(s):
    if len(s) < 3:
        return s
    elif s.endswith("ing"):
        return s + "ly"
    else:
        return s + "ing"

print(add_string("play"))
print(add_string("playing"))
print(add_string("go"))
```

```python
# 22) Reverse string if length multiple of 4
def reverse_if_multiple_of4(s):
    if len(s) % 4 == 0:
        return s[::-1]
    return s

print(reverse_if_multiple_of4("abcd"))
print(reverse_if_multiple_of4("hello"))
```

```python
# 23) String with first 2 + last 2 chars
def first_last_two(s):
    if len(s) < 2:
        return ""
    return s[:2] + s[-2:]

print(first_last_two("python"))
print(first_last_two("p"))
```

```python
# 31) Count strings with same first and last char (length ≥ 2)
def match_words(words):
    count = 0
    for word in words:
        if len(word) >= 2 and word[0] == word[-1]:
            count += 1
    return count

print(match_words(['abc', 'xyz', 'aba', '1221']))
```

```python
# 63) Check if number is Perfect

 # Perfect number = sum of divisors (excluding itself) = number
# Example: 6 → divisors 1+2+3=6

def is_perfect(n):
    if n < 2:
        return False
    divisors_sum = sum(i for i in range(1, n) if n % i == 0)
    return divisors_sum == n

print(is_perfect(6))   # True
print(is_perfect(28))  # True
print(is_perfect(12))
```

```python
# 83) Exception handling in Python

Exception Handling: Process of responding to runtime errors in a controlled way using try, except.

Error: An error is a problem in a program that prevents execution. Types:

SyntaxError → wrong syntax

ZeroDivisionError → divide by zero

TypeError → wrong type usage
```

```python
# 84) How many except statements can a try-except block have?

Any number, multiple except blocks allowed.

Some built-in exceptions: ValueError, IndexError, KeyError, ZeroDivisionError, FileNotFoundError.
```

```python
# 85) When is else part executed?

# The else block executes if no exception occurs in the try block.
try:
    print(10 / 2)
except ZeroDivisionError:
    print("Error")
else:
    print("No error, else executed")
```

```python
# 86) Can one except handle multiple exceptions?

# Yes, using a tuple:

try:
    x = int("abc")
except (ValueError, TypeError) as e:
    print("Error:", e)
```

```python
# 87) When is finally block executed?

# Always executed regardless of exception, typically used for cleanup.

try:
    x = 10 / 2
finally:
    print("This always runs")
```
